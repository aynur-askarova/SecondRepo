Grading System

```python
exam_marks=float(input('Exam marks: '))
if exam_marks>=90:
    print('Grade: A')
elif 90>exam_marks>=75:
    print('Grade: B')
elif 75>exam_marks>=50:
    print('Grade: C')
else:
    print('Grade: F')
```

Multiplication Table

```python
number=int(input('Enter a number: '))
for i in range(1,11):
    print(f'{number} x {i} = {number*i}')
```

Password Retry System


```python
correct_password='Python123'
attempt=0
k=False
while attempt<3:
    password=input('Enter the password: ')
    if password==correct_password:
        print('Access')
        k=True
        break
    else:
        print('Incorrect password')
        attempt+=1
if k==False:
    print('All attempts failed. Access denied.')
```
