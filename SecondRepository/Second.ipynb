Task 1

```python
student_score=float(input("Enter student score: "))
if 91<=student_score<=100:
    print('A')
elif 81<=student_score<=90:
    print('B')
elif 71<=student_score<=80:
    print('C')
elif 61<=student_score<=70:
    print('D')
elif 51<=student_score<=60:
    print('E')
else:
    print('F')
```

Task 2

```python
n=int(input('Enter positive integer: '))
s=0
k=0
for i in range(1,n):
    if i%2==0:
        s+=i
    if i%15==0:
        k+=1
        print(' 3 5')
    elif i%3==0:
        k+=1
        print('3')
    elif i%5==0:
        print('5')
    else:
        print(n)
print(f'The sum of all even numbers from 1 to {n} is {s}')
print(f'{k} numbers were divisible by 3')
```
